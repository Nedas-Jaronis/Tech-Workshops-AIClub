# How do language models behave?

- Models predict outputs from learned patterns and context.
- Different runs can return different answers.
- Fluent text is not proof that a claim is true.

**Principle:** A good prompt improves clarity; verification still matters.


# The five-part prompt framework

1. **Task** — what to do
2. **Context** — relevant background/data
3. **Constraints** — limits and boundaries
4. **Examples** — show the desired pattern
5. **Output format** — what success looks like


# Make the prompt testable

**Before:** `Classify this ticket.`

**After:** `Classify this support ticket as Billing, Technical, Account Access, or Other. Reply with only one category: "I was charged twice."`

**Discuss:** What changed? How can we score the answer?


# Zero-shot, one-shot, few-shot

**Zero-shot:** instructions only

**One-shot:** one labeled example

**Few-shot:** two or more labeled examples

**Rule:** More examples do not always mean better results. Test them.

# Make iteration measurable

1. Choose a failure case
2. Make one specific prompt change
3. Rerun the same held-out tickets
4. Explain what improved (and what didn't)

**Reminder:** Five examples are for demonstration, not production evidence.


# Context and reliability

- Supply relevant source material when facts matter.
- Ask for uncertainty when evidence is insufficient.
- Verify citations and critical claims externally.
- A large context window is not a guarantee of correctness.


# Beyond a basic prompt

**RAG:** retrieve evidence before answering

**Tool calling:** a model can request a calculator or other tool; the application decides what to execute

**Prompt injection:** untrusted documents must not become trusted instructions

These are optional follow-on topics.


# Exit challenge (3 minutes)

Rewrite: `Help me study machine learning.`

Include a specific task, learner context, constraints, example or reference material, and desired output format.

**Question:** How would you test whether it worked?


# Takeaways

**Be specific → provide context → set format → test → revise**

A prompt is not finished when it *looks* good. It is finished when it performs well enough on representative examples.
